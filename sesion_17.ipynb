# Notas: Romper la Barrera del No (Chris Voss)

Estas son notas estructuradas basadas en las técnicas de negociación de rehenes del FBI aplicadas a la vida diaria y los negocios.

### 1. Rehenes vs. Harvard (Chris Voss)
**Concepto:** La negociación tradicional (estilo Harvard) asume que las personas son racionales y actúan por intereses lógicos. Sin embargo, Chris Voss descubrió en el FBI que las negociaciones en situaciones de rehenes están impulsadas por las emociones. Los humanos somos seres emocionales e irracionales en situaciones de tensión.

**Ejemplo:**
* **Enfoque Lógico (Fallido):** "Si no bajas el precio en un 10%, perderás la venta. Es pura matemática."
* **Enfoque Emocional (Voss):** Abordar los miedos y deseos del otro antes de hablar de números, creando un ambiente donde la persona se sienta segura para ceder.

### 2. El arma secreta de crisis (La Empatía Táctica)
**Concepto:** En la interacción humana durante una crisis, tu arma más fuerte no es la intimidación, sino la **empatía táctica**. Significa escuchar activamente y demostrarle a la otra persona que comprendes profundamente su perspectiva, sentimientos y la situación en la que se encuentra, incluso si no estás de acuerdo con ella.

**Ejemplo:**
* **Técnica del Etiquetado (Labeling):** En lugar de decir "Te entiendo", dices: *"Parece que estás muy frustrado con los plazos de este proyecto"*. Al nombrar la emoción, ayudas a desactivar la negatividad del otro.

### 3. El truco mental Jedi (El Espejeo / Mirroring)
**Concepto:** Consiste en repetir, en un tono de voz calmado, las últimas 1 a 3 palabras clave que la otra persona acaba de decir. Es como un truco Jedi porque la gente siente la necesidad subconsciente de explicar y elaborar más sobre lo que acaban de decir. Esto genera conexión y te da más información sin necesidad de presionar.

**Ejemplo (Jefe y Empleado):**
* **Jefe:** "Necesito que saques copias de todos estos informes para esta tarde, no tenemos tiempo."
* **Empleado:** "¿No tenemos tiempo?" *(Pausa en silencio)*
* **Jefe:** "Sí, es que el cliente adelantó la reunión para mañana por la mañana y necesitamos tener todo listo hoy..."

*(Resultado: El empleado obtuvo el contexto real y transformó una orden en una conversación sin negarse abiertamente).*.

### 4. El poder del "Cómo" (Preguntas Calibradas)
**Concepto:** Usar preguntas abiertas que comiencen con **"¿Cómo...?"** o **"¿Qué...?"**. Estas preguntas piden ayuda al oponente, dándole la ilusión de que tiene el control de la conversación, mientras que en realidad tú estás guiando la negociación. Fuerza a la otra parte a detenerse y pensar en resolver *tu* problema.

**Ejemplo:**
* **Exigencia irracional:** "¡Quiero que me pagues el doble por este trabajo!"
* **Pregunta calibrada:** *"¿Cómo se supone que voy a hacer eso?"*
*(Esto obliga a la otra persona a ponerse en tus zapatos y considerar tus limitaciones, a menudo bajando sus propias demandas).*.

### 5. La mentalidad correcta (Negociación como descubrimiento)
**Concepto:** La negociación no es una batalla de argumentos donde alguien tiene que ganar o perder, ni una competencia para aplastar al otro. La misión es entrar a la conversación como si fuera un proceso de descubrimiento. Debes asumir que hay información crítica que desconoces (los "Cisnes Negros").

**La conclusión clave:** Adopta una mentalidad de curiosidad. Tu objetivo es descubrir qué es lo que realmente valora la otra persona y cuáles son los miedos ocultos que están bloqueando un acuerdo para encontrar una verdadera solución.

# Mapa Legal de Software en Colombia

Guía de aspectos legales, contractuales y de protección para el desarrollo e implementación de proyectos de software en el ecosistema digital colombiano.

### 1. El contrato como guía (Límites exactos y entregables)
**Concepto:** Todo proyecto de software necesita reglas claras desde el principio. El contrato es el mapa de ruta y debe tener límites exactos para evitar la "corrupción del alcance" (scope creep).

**Elementos clave:**
* **Entregables específicos:** Definir exactamente qué se va a entregar, en qué formato y en qué fechas.
* **Ejemplo:** En lugar de poner "Desarrollo de una app de ventas", se debe detallar: "Módulo de registro, pasarela de pagos integrada con PayU, y panel de administrador, entregables en 3 fases de 4 semanas cada una".

### 2. Leyes del Ecosistema Digital en Colombia
**Concepto:** El desarrollo de software está regulado por normas específicas que protegen la creación, la transacción y la información.

**Normativa principal:**
* **Ley 23 de 1982 (Derechos de Autor):** Protege el código fuente y la arquitectura del software como obra literaria.
* **Ley 527 de 1999 (Comercio Electrónico):** Regula los mensajes de datos, firmas digitales y contratos electrónicos, dándoles validez jurídica.
* **Ley 1581 de 2012 (Protección de Datos Personales - Habeas Data):** Obliga a implementar políticas de privacidad si el software recolecta datos de usuarios.

**Conceptos de responsabilidad:**
* **Daño Emergente:** El perjuicio o la pérdida directa que sufre una parte (ej. costo de servidores inútiles por software defectuoso).
* **Lucro Cesante:** El dinero que se dejó de ganar por culpa del problema (ej. ventas perdidas porque la app se cayó 3 días).

### 3. Incumplimiento del proyecto
**Concepto:** ¿Qué pasa si el software no hace lo que prometía o no se entrega a tiempo? Se deben establecer cláusulas claras para gestionar el fracaso del proyecto.

**Elementos a considerar:**
* Definir qué constituye un "incumplimiento grave" versus un "bug menor".
* **Cláusula penal:** Acordar multas o penalidades por retrasos injustificados.
* **Mecanismos de salida:** Cómo se terminará la relación si el proyecto es inviable, incluyendo quién se queda con el código desarrollado hasta el momento.

### 4. Seguros y Escudos Protectores
**Concepto:** Los errores en el software pueden costar muy caro. Es vital contar con pólizas de seguro (como el Seguro de Responsabilidad Civil Profesional - RCP) para proteger el patrimonio de la empresa desarrolladora.

**Coberturas comunes:** Errores u omisiones en el código, vulneraciones a la privacidad o fuga de datos.

**Exclusiones (Lo que NO cubre la póliza):**
* **Mala fe:** Actos fraudulentos o código malicioso intencional.
* **Retiro de soporte:** Fallos que ocurren porque se dejó de dar mantenimiento al software de forma intencionada.
* **Pruebas Beta:** Errores en versiones de software que no están en fase de producción definitiva.

**Amparo de Cumplimiento:** Una póliza exigida a menudo en contratos públicos o con grandes empresas que garantiza que si el desarrollador incumple, la aseguradora indemniza al cliente.

### 5. Checklist antes de la firma
**Concepto:** Antes de firmar cualquier contrato de desarrollo o cesión de derechos de software, es obligatorio revisar una lista de puntos críticos.

**Verificar en la lista:**
1. **Alcance:** ¿Está 100% claro qué incluye y qué NO incluye el proyecto?
2. **Propiedad Intelectual:** ¿Quién es el dueño del código fuente y los derechos patrimoniales una vez terminado y pagado el proyecto?
3. **Confidencialidad (NDA):** ¿Están protegidos los secretos comerciales de ambas partes?
4. **Garantía y Soporte:** ¿Por cuánto tiempo se arreglarán los bugs sin costo adicional tras la entrega?
