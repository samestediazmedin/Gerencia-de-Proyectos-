## **Valor Ganado (Earned Value Management - EVM)**

El Valor Ganado es una metodología de gestión de proyectos que integra el alcance, el cronograma y los costos para evaluar el rendimiento y el progreso del proyecto de manera objetiva. Permite a los gerentes de proyecto medir la salud del proyecto en tiempo real.

### 1. El diagnóstico (1.0 - proporcional)

El diagnóstico en el contexto del Valor Ganado se refiere a la evaluación del desempeño del proyecto utilizando métricas clave. El valor `1.0` (o 100%) es el punto de referencia ideal, indicando que el proyecto está exactamente en el cronograma y dentro del presupuesto. Los valores por encima o por debajo de 1.0 son proporcionales al desvío.

*   **Índice de Desempeño del Costo (CPI - Cost Performance Index)**: Mide la eficiencia del costo. Un CPI de 1.0 significa que el proyecto está gastando lo esperado para el trabajo realizado. Si el CPI es 0.8, significa que por cada dólar planeado, solo se ha logrado un valor de 80 centavos de trabajo.
    *   **Ejemplo**: Si un proyecto tiene un CPI de `0.85`, significa que por cada peso que se esperaba gastar para el trabajo realizado, se está gastando más, o en otras palabras, el valor ganado es solo el 85% del costo real incurrido. El diagnóstico es que el proyecto está por encima del presupuesto.

*   **Índice de Desempeño del Cronograma (SPI - Schedule Performance Index)**: Mide la eficiencia del cronograma. Un SPI de 1.0 significa que el proyecto está en el cronograma. Si el SPI es 1.2, significa que se está avanzando un 20% más rápido de lo planificado. Si es 0.9, se está avanzando un 10% más lento.
    *   **Ejemplo**: Un proyecto con un SPI de `0.9` indica que se está completando el 90% del trabajo que debería haberse hecho según el cronograma. El diagnóstico es que el proyecto está atrasado.

### 2. Un buen gerente sabe dónde está

Un buen gerente de proyecto utiliza el Valor Ganado para tener una comprensión clara y objetiva del estado actual del proyecto en términos de costo y cronograma. No se basa en suposiciones, sino en datos concretos.

*   **Ejemplo**: Un gerente de proyecto revisa los informes de EVM y ve que el CPI es `0.95` y el SPI es `1.05`. Inmediatamente sabe que el proyecto está ligeramente por encima del presupuesto (5% más caro de lo previsto para el trabajo realizado) pero un poco adelantado en el cronograma (5% más rápido de lo esperado). Esta información le permite tomar decisiones proactivas, como investigar las causas del sobrecosto y quizás reasignar recursos para mantener el proyecto en línea.

### 3. El esfuerzo requerido (TCPI - Graficar el escenario)

El **TCPI (To Complete Performance Index)** es una proyección del rendimiento futuro requerido para cumplir con el presupuesto restante. Ayuda a determinar qué tan eficiente debe ser el equipo en el trabajo restante para terminar dentro del presupuesto total. Graficar el escenario implica visualizar el impacto de diferentes TCPI en la finalización del proyecto.

*   **Fórmula**: TCPI = (BAC - EV) / (BAC - AC)
    *   BAC: Budget at Completion (Presupuesto al Finalizar)
    *   EV: Earned Value (Valor Ganado)
    *   AC: Actual Cost (Costo Real)

*   **Interpretación**: Un TCPI de `1.0` significa que el equipo debe continuar trabajando con la misma eficiencia actual para terminar dentro del presupuesto. Un TCPI de `1.2` significa que el equipo debe ser un 20% más eficiente de lo que ha sido hasta ahora para cumplir con el presupuesto restante, lo cual puede ser un desafío. Un TCPI menor a `1.0` sugiere que tienen margen para ser menos eficientes.

*   **Ejemplo**: Un proyecto tiene un BAC de $1,000,000. Hasta la fecha, el EV es $400,000 y el AC es $450,000.
    *   TCPI = (1,000,000 - 400,000) / (1,000,000 - 450,000) = 600,000 / 550,000 ≈ `1.09`
    *   **Significado**: El equipo debe lograr un 9% más de eficiencia en el trabajo restante de lo que se había planeado inicialmente para terminar dentro del presupuesto total. Este es un desafío, y el gerente podría **graficar este escenario** para mostrar al equipo que se requiere un esfuerzo adicional significativo o considerar una revisión del presupuesto.

## **Consultores de Élite: Gestión de Riesgos**

La gestión de riesgos es un componente crítico de la gestión de proyectos, donde los consultores de élite aplican un enfoque estructurado para identificar, analizar y responder a las incertidumbres que pueden afectar los objetivos del proyecto.

### 1. Introducción a los riesgos

Un riesgo es un evento o condición incierta que, si ocurre, tiene un efecto positivo o negativo en uno o más objetivos del proyecto (como el alcance, el cronograma, el costo o la calidad). La introducción a los riesgos implica establecer la importancia de su gestión, definir la terminología clave y comprender el ciclo de vida de la gestión de riesgos.

*   **Ejemplo**: Al inicio de un proyecto de desarrollo de software, se introduce el concepto de riesgos explicando que un "riesgo" no es un problema actual, sino un evento futuro potencial. Se identifican riesgos iniciales como "la falta de experiencia del equipo con una nueva tecnología" (negativo) o "la posibilidad de reutilizar código existente de otro proyecto" (positivo).

### 2. Análisis cualitativo de riesgo (Matriz de Riesgos)

El análisis cualitativo de riesgos evalúa la probabilidad de ocurrencia de un riesgo y su impacto potencial en los objetivos del proyecto, utilizando escalas descriptivas (ej. alta, media, baja) para priorizar los riesgos. La matriz de riesgos es una herramienta visual clave que mapea la probabilidad y el impacto para asignar un nivel de riesgo general (ej. crítico, alto, moderado, bajo).

*   **Ejemplo**: Para un proyecto de construcción, se identifica el riesgo de "retraso en la entrega de materiales clave".
    *   **Probabilidad**: Se estima "Media".
    *   **Impacto**: Se estima "Alto" (porque detendría el progreso de la obra).
    *   En la matriz de riesgos, un riesgo "Media-Alto" podría clasificarse como "Riesgo Alto", lo que indica que debe ser abordado con prioridad sobre otros riesgos "Bajos".

### 3. Análisis cuantitativo (Árbol de decisiones, simulación de Monte Carlo)

El análisis cuantitativo de riesgos va más allá de la priorización y asigna valores numéricos a la probabilidad y el impacto de los riesgos. Se utiliza para evaluar el efecto global de los riesgos en los objetivos del proyecto y para tomar decisiones bajo incertidumbre.

*   **Árbol de Decisiones**: Una herramienta gráfica que ayuda a visualizar las posibles consecuencias de diferentes decisiones y los riesgos asociados. Cada rama representa una posible decisión o un evento de riesgo.
    *   **Ejemplo**: Una empresa debe decidir entre dos proveedores para un componente crítico. El **árbol de decisiones** podría modelar el costo de cada proveedor, la probabilidad de defectos en sus componentes y el costo de reprocesamiento o retraso si se produce un defecto. Esto permite calcular el valor monetario esperado de cada ruta y elegir la opción óptima.

*   **Simulación de Monte Carlo**: Una técnica que utiliza modelos informáticos para simular los resultados de un proyecto muchas veces, teniendo en cuenta la incertidumbre de variables como el costo o la duración de las actividades. Genera un rango de posibles resultados y sus probabilidades.
    *   **Ejemplo**: En un proyecto de desarrollo de un nuevo producto, se pueden estimar rangos de tiempo y costo para cada fase (ej. diseño: 5-8 semanas, 50,000-70,000 USD). La **simulación de Monte Carlo** ejecutaría miles de iteraciones, seleccionando aleatoriamente valores dentro de esos rangos para cada fase, y al final, proporcionaría una distribución de probabilidad para la duración total y el costo total del proyecto (ej. "Hay un 80% de probabilidad de que el proyecto termine en 18-22 semanas y cueste entre 200,000 y 250,000 USD").

### 4. Análisis de los resultados (Identificar: asignar - valores - multiplicar)

Después de realizar los análisis cualitativo y cuantitativo, los consultores de élite se centran en el análisis profundo de los resultados. Esto implica comprender las implicaciones de los riesgos identificados y cuantificados.

*   **Identificar**: Reconocer los riesgos más críticos y aquellos con mayor impacto potencial. Esto se desprende directamente de la matriz de riesgos o de los resultados de la simulación de Monte Carlo.
*   **Asignar**: Atribuir responsabilidades para la gestión de cada riesgo. ¿Quién será el propietario del riesgo y el encargado de monitorearlo y ejecutar la respuesta?
*   **Valores**: Entender los valores numéricos asociados a los riesgos, ya sean probabilidades e impactos cualitativos o rangos de costos y tiempos cuantificados. Esto permite una discusión basada en datos.
*   **Multiplicar**: En el contexto del análisis cualitativo, es la 'multiplicación' (o combinación) de probabilidad por impacto para obtener un nivel de riesgo. En el análisis cuantitativo, es la consideración de cómo múltiples riesgos pueden interactuar y "multiplicar" sus efectos en el resultado global del proyecto.

*   **Ejemplo**: Después de una simulación de Monte Carlo, se **identifica** que el riesgo de "fluctuación del tipo de cambio" tiene un 60% de probabilidad de aumentar el costo del proyecto en un 5-10%. Se **asigna** la responsabilidad de monitorear este riesgo al departamento financiero. Se revisan los **valores** (probabilidad, impacto en el costo). Se comprende cómo este riesgo puede **multiplicarse** con otros riesgos de la cadena de suministro para impactar el presupuesto final del proyecto.

### 5. Estrategia de respuesta a los riesgos

Una vez que los riesgos han sido analizados, se desarrollan planes para responder a ellos. Las estrategias comunes incluyen:

*   **Evitar**: Eliminar la causa del riesgo (ej. cambiar a un proveedor más confiable).
*   **Transferir**: Trasladar el impacto del riesgo a un tercero (ej. comprar un seguro, subcontratar).
*   **Mitigar**: Reducir la probabilidad o el impacto de un riesgo (ej. realizar pruebas adicionales, implementar planes de contingencia).
*   **Aceptar**: No hacer nada y aceptar el impacto si el riesgo ocurre (generalmente para riesgos de bajo impacto o probabilidad, o donde el costo de la respuesta es mayor que el riesgo en sí).
*   **Explotar (para riesgos positivos)**: Tomar medidas para asegurar que la oportunidad ocurra (ej. asignar recursos adicionales para aprovechar una nueva tecnología).
*   **Mejorar (para riesgos positivos)**: Aumentar la probabilidad o el impacto de una oportunidad (ej. realizar una inversión inicial para mejorar la eficiencia).
*   **Compartir (para riesgos positivos)**: Formar asociaciones para maximizar la oportunidad.

*   **Ejemplo**: Para el riesgo de "retraso en la entrega de materiales clave" (identificado en el punto 2):
    *   **Mitigar**: Establecer un proveedor de respaldo con un acuerdo de entrega rápida, incluso si es un poco más caro.
    *   **Aceptar**: Si el material no es crítico para el camino crítico y el retraso es menor, la empresa podría aceptar el pequeño impacto.
    *   **Transferir**: Negociar un contrato con el proveedor principal que incluya penalizaciones por retrasos.

### 6. La creación de registros de riesgos

El registro de riesgos (Risk Register) es un documento fundamental que detalla todos los riesgos identificados en el proyecto. Actúa como un repositorio central para la información de cada riesgo, facilitando su seguimiento y gestión a lo largo del ciclo de vida del proyecto.

Un registro de riesgos típico incluye:
*   **ID del Riesgo**: Un identificador único.
*   **Descripción del Riesgo**: Una declaración clara del evento de riesgo.
*   **Causa del Riesgo**: Lo que podría llevar a que ocurra el riesgo.
*   **Impacto del Riesgo**: Las consecuencias si el riesgo ocurre.
*   **Probabilidad e Impacto**: Cualitativos (ej. alto, medio, bajo) o cuantitativos (ej. 70% de probabilidad de un impacto de $50,000).
*   **Puntuación de Riesgo**: Calculada a partir de la probabilidad y el impacto.
*   **Propietario del Riesgo**: La persona responsable de monitorear el riesgo y ejecutar la respuesta.
*   **Estrategia de Respuesta**: La acción planificada para abordar el riesgo (evitar, mitigar, etc.).
*   **Estado del Riesgo**: Abierto, cerrado, en monitoreo.

*   **Ejemplo**: Creación de una entrada en el registro de riesgos:
    *   **ID**: R001
    *   **Descripción**: Posible escasez de mano de obra calificada en el mercado local.
    *   **Causa**: Alta demanda de habilidades específicas en la industria.
    *   **Impacto**: Retraso en el cronograma y aumento de costos debido a la contratación de personal externo o capacitación.
    *   **Probabilidad**: Media
    *   **Impacto (cualitativo)**: Alto
    *   **Puntuación de Riesgo**: Alto
    *   **Propietario**: Gerente de Recursos Humanos
    *   **Estrategia de Respuesta**: Mitigar (iniciar un programa de capacitación interno temprano; explorar asociaciones con escuelas técnicas).
    *   **Estado**: Abierto - En Monitoreo

Este registro se revisa y actualiza regularmente durante las reuniones del proyecto.
