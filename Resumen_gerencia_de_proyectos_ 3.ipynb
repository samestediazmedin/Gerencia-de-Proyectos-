## Contexto Teórico y Explicaciones Académicas

Para facilitar tu aprendizaje, aquí tienes el desglose conceptual detallado de cada punto de tus notas:

### 1. ¿Qué es la EDT (Estructura de Desglose de Trabajo o WBS)?
La **EDT** es una herramienta fundamental en la gestión de proyectos. Consiste en dividir de forma jerárquica el trabajo total del proyecto en componentes más pequeños y manejables (llamados paquetes de trabajo). Si un trabajo no está en la EDT, queda fuera del alcance del proyecto.
* **Diccionario de la EDT:** Es el documento que acompaña al diagrama y describe detalladamente cada componente de la EDT (recursos necesarios, responsable, criterios de aceptación, duración estimada, etc.).

### 2. La Regla del 100%
Es la regla de oro para construir una EDT:
* Establece que el trabajo de los niveles inferiores (hijos) debe sumar exactamente el 100% del trabajo representado por el nivel superior (padre).
* No se puede realizar trabajo adicional fuera de lo estipulado (evita el *scope creep* o corrupción del alcance).
* Tampoco se puede omitir ninguna tarea necesaria para completar el entregable.

### 3. EDV (Estructura de Desglose de Valor)
A diferencia de la EDT tradicional (que se enfoca estrictamente en tareas y entregables físicos), la **EDV** vincula cada entregable del proyecto con el **valor de negocio** real que va a aportar a la organización. Asegura que el esfuerzo técnico se traduzca directamente en beneficios tangibles o estratégicos para la empresa.

### 4. Descomposición Modular de Software
Al gestionar un proyecto de software, se aplica una descomposición modular para no perder el control del desarrollo. Un proyecto típico se subdivide en módulos funcionales y áreas clave de soporte, tales como:
* **Administración del sistema:** Gestión de usuarios, roles y seguridad.
* **Documentación técnica y de usuario:** Manuales, arquitecturas y guías.
* **Programa de capacitación:** Transferencia de conocimiento a los usuarios finales para asegurar la adopción del software.
