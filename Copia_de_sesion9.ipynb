# **4. ¿Cuándo se puede instalar el servidor?**

Esta es una de las preguntas importantes.

En nuestro ejemplo, la instalación se realiza en el día 25.

Pero no significa que el servidor aparezca ese día.

Antes tenemos:

Día 20-21 → Preparar servidor

Después:

Día 23-24 → Integrar sistema

Y finalmente:

Día 25 → Instalar software en servidor

Entonces podemos explicarlo así:

“El servidor se prepara antes de la instalación. La instalación está programada para el día 25, después de completar la integración del software y verificar que el servidor esté listo.”

¿Por qué no instalarlo desde el principio?

Porque todavía no tenemos el software terminado.

Sería como intentar poner un programa en producción cuando todavía estamos diseñándolo.

# **6.  Dependencias del proyecto**

Dependencia obligatoria

El software debe estar integrado antes de instalarlo.

Integrar sistema → Instalar sistema.

No podemos saltarnos ese orden.

Dependencia discrecional

El equipo decide que primero se haga el módulo de usuarios y después el módulo de mediciones.

Podría organizarse de otra forma, pero el equipo escogió ese orden.

Dependencia interna

El programador necesita que el diseñador entregue las pantallas.

Dependencia externa

El servidor puede depender de:

Proveedor de hosting.
Administrador externo.
Servicio de internet.
Certificado SSL.
Servicio de correo.

Si el proveedor no entrega el servidor a tiempo, puede afectar la fecha de implementación.

# **7.  Adelanto y  retraso**

Ahora podemos poner ejemplos reales dentro del proyecto.

Adelanto

Supongamos que la preparación del servidor termina antes de lo esperado.

Estaba planeada para terminar el día 21.

Pero termina el día 20.

Entonces podemos adelantar algunas actividades relacionadas.

Adelanto = ganar tiempo.

Retraso

Supongamos que el servidor debía estar listo el día 21, pero el proveedor tiene un problema y lo entrega el día 23.

Tenemos:

Retraso = 2 días.

Eso podría provocar que la instalación pase del:

Día 25 → Día 27

Y si afecta las pruebas, también debemos moverlas.

Esto es justamente lo que significa ajustar el cronograma.

# **10.  ¿Qué pasa si aparecen muchos errores?**

Aquí entra el control del proyecto.

Por ejemplo:

El plan dice:

Correcciones = 2 días.

Pero aparecen demasiados errores y necesitamos 4 días.

Entonces:

Plan original:

Pruebas → 2 días
Correcciones → 2 días
Entrega → día 30

Situación real:

Pruebas → 2 días
Correcciones → 4 días
Entrega → día 32

Tenemos un retraso de 2 días.

El director del proyecto debe revisar:

Qué errores aparecieron.
Cuánto tiempo requieren.
Cuánto dinero adicional cuestan.
Si afectan la fecha de entrega.
Si hay actividades que pueden adelantarse.
Si es necesario modificar el cronograma.

# **14. 🔄 Explicacion del proyecto: **

Primero definimos las actividades que necesitamos para desarrollar el software. Después las secuenciamos según sus dependencias. Luego estimamos cuánto tiempo y dinero necesitamos para cada actividad. Durante el proyecto hacemos seguimiento y, si aparecen retrasos o errores, ajustamos el cronograma.


En nuestro proyecto, primero se levantan los requisitos, después se diseña el sistema y la base de datos, luego se programa. Mientras se termina el desarrollo también se puede preparar el servidor. Una vez que el software está integrado y el servidor está preparado, se realiza la instalación en el día 25. Posteriormente se hacen pruebas en el servidor, se corrigen los errores y finalmente se entrega el proyecto en el día 30.

Los errores se registran durante las pruebas, se estima el tiempo y costo de corregirlos y se revisa su impacto en el cronograma. Si las correcciones generan un retraso, se ajustan las fechas de las actividades siguientes. En nuestro ejemplo tenemos un costo estimado de $645.000 COP para correcciones.


Porque antes necesitamos terminar la programación, preparar el servidor e integrar el sistema. Por eso la instalación depende de esas actividades y se establece como un hito del proyecto.
