# 1. ¿Qué es TCP/IP y por qué es fundamental?
El modelo **TCP/IP** (Transmission Control Protocol / Internet Protocol) es la base de todo el funcionamiento del Internet actual.
* **TCP (Protocolo de Control de Transmisión):** Se encarga de dividir la información en pequeños paquetes antes de enviarla, y se asegura de volver a unirlos correctamente en el destino. Si un paquete se pierde en el camino, TCP solicita que se vuelva a enviar para garantizar que no haya pérdidas.
* **IP (Protocolo de Internet):** Se encarga de direccionar y enrutar los paquetes para que viajen a través de la red y lleguen al destino correcto (utilizando las direcciones IP de origen y destino).

### 2. El Modelo OSI (Open Systems Interconnection)
Es un modelo de referencia conceptual creado por la ISO para permitir que diferentes sistemas de comunicación se conecten de manera estándar. Se divide en **7 capas**:
1. **Capa Física:** Transmisión de bits a través de cables o señales de radio.
2. **Capa de Enlace de Datos:** Direccionamiento físico (direcciones MAC) y detección de errores.
3. **Capa de Red:** Determinación de la mejor ruta física para enviar los datos (Direcciones IP, Routers).
4. **Capa de Transporte:** Conexión extremo a extremo y confiabilidad del envío (TCP y UDP).
5. **Capa de Sesión:** Mantenimiento y control de la sesión entre las aplicaciones que se comunican.
6. **Capa de Presentación:** Traducción del formato de los datos, compresión y cifrado (asegura que el receptor entienda el contenido).
7. **Capa de Aplicación:** Interacción directa con el usuario final (Protocolos como HTTP, FTP, SMTP).

# 3. Seguridad en Redes y su Importancia
* **Cifrado (Criptografía):** Consiste en alterar la información original mediante algoritmos matemáticos para que solo las personas autorizadas (con la clave de descifrado) puedan entenderla. Esto es vital para transacciones bancarias, contraseñas y privacidad en línea.
* **Firewall (Cortafuegos):** Es una barrera de seguridad que monitorea y filtra el tráfico de red entrante y saliente según reglas establecidas, bloqueando accesos no autorizados.
